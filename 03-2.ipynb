```python
import torch
import transformers

print("Torch :", torch.__version__)
print("Transformers :", transformers.__version__)


from transformers import PreTrainedTokenizerFast
from transformers import BartForConditionalGeneration


MODEL_NAME = "gogamza/kobart-summarization"

tokenizer = PreTrainedTokenizerFast.from_pretrained(MODEL_NAME)

model = BartForConditionalGeneration.from_pretrained(MODEL_NAME)

print("KoBART 모델 로드 완료")


text = """
인공지능 기술은 최근 다양한 산업 분야에서 빠르게 발전하고 있다.
특히 생성형 AI는 문서 작성, 번역, 코드 생성, 이미지 생성 등 여러 분야에서 활용되고 있다.
기업들은 업무 자동화를 위해 AI를 적극 도입하고 있으며,
교육 분야에서도 AI를 활용한 맞춤형 학습 서비스가 확대되고 있다.
그러나 개인정보 보호와 AI 윤리 문제에 대한 사회적 논의도 함께 이루어지고 있다.
"""

inputs = tokenizer(
    text,
    return_tensors="pt",  ## 결과를 PyTorch Tensor 형태로 변환
    truncation=True,
    max_length=1024
)

print(inputs.keys())

#토큰 아이디와 attention_mask 를 반환
#토큰 아이디 : 어휘 사전에서 가져온 정수 값
#attention_mask : 1, 0(패딩처리)


summary_ids = model.generate(
    inputs["input_ids"],
    attention_mask=inputs["attention_mask"],

    max_length=64,
    min_length=20,

    num_beams=4,
    # 단어 후보

    no_repeat_ngram_size=2, #2번이상 반복 불가

    early_stopping=True
)


summary = tokenizer.decode(
    summary_ids[0],
    skip_special_tokens=True
)

print("="*60)
print("원문")
print("="*60)

print(text)

print()

print("="*60)
print("요약")
print("="*60)

print(summary)
```
